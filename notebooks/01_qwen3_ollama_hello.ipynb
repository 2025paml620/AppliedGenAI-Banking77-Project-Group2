# Step 1: Call a local Qwen3 1.7B model

Goal: confirm the local model answers a banking query.

Requirements: Ollama running on `localhost:11434` with `qwen3:1.7b` pulled (`ollama pull qwen3:1.7b`).
Standard library only, so no extra packages are needed.

```python
import json
import urllib.request

OLLAMA_URL = "http://localhost:11434/api/chat"
MODEL = "qwen3:1.7b"


def chat(messages, temperature=0.0, max_tokens=256):
    """Send chat messages to the local Ollama server and return the reply text."""
    payload = {
        "model": MODEL,
        "messages": messages,
        "stream": False,
        "think": False,  # skip Qwen3's reasoning block so the reply is the answer
        "options": {"temperature": temperature, "num_predict": max_tokens},
    }
    req = urllib.request.Request(
        OLLAMA_URL,
        data=json.dumps(payload).encode("utf-8"),
        headers={"Content-Type": "application/json"},
    )
    with urllib.request.urlopen(req, timeout=120) as resp:
        body = json.loads(resp.read().decode("utf-8"))
    return body["message"]["content"].strip()


reply = chat([
    {"role": "system", "content": "You are a banking customer service assistant."},
    {"role": "user", "content": "How do I activate my new credit card?"},
])
print(reply)
```
